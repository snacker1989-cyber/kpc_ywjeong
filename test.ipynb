```python
from RAGbuilder import load_and_split

def debug_chunks(pdf_path):
    # 1. 문서 로드 및 분할
    docs = load_and_split(pdf_path)
    
    print(f"📄 파일명: {pdf_path}")
    print(f"✂️ 총 생성된 청크 수: {len(docs)}개")
    print("-" * 50)

    # 2. 상위 5개 청크 상세 확인
    for i, doc in enumerate(docs[:5]):  # 처음 5개만 확인
        content = doc.page_content.replace("\n", " ") # 가독성을 위해 줄바꿈 제거
        print(f"[Chunk {i}]")
        print(f"📏 길이: {len(doc.page_content)}자")
        print(f"📝 내용: {content[:200]}...") # 앞부분 200자만 출력
        
        # 중첩(Overlap) 확인을 위해 다음 청크의 시작 부분 살짝 보여주기
        if i < len(docs) - 1:
            next_start = docs[i+1].page_content[:30].replace("\n", " ")
            print(f"🔗 다음 청크와의 연결부: ... [{next_start}]")
        
        print("-" * 50)
```

```python
# 실행 예시
if __name__ == "__main__":
    test_pdf = "data/(4-01) 취업규칙(250123).pdf"
    debug_chunks(test_pdf)
```
